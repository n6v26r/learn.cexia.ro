## 1. Căutarea liniară

Parcurgem elementele de la stânga la dreapta și ne oprim când găsim ținta. Algoritmul funcționează pentru orice listă, inclusiv pentru una neordonată. În cel mai nefavorabil caz inspectează toate cele $n$ elemente.

```python
def cautare_liniara(valori, tinta):
    comparatii = 0
    for indice, valoare in enumerate(valori):
        comparatii += 1
        if valoare == tinta:
            return indice, comparatii
    return -1, comparatii

date = [14, 3, 19, 8, 2, 11, 7, 5]
pozitie, pasi = cautare_liniara(date, 7)
print("Poziție:", pozitie)
print("Comparații:", pasi)
```

### Invariantul buclei

Înaintea fiecărei iterații știm că ținta nu apare în porțiunea deja parcursă. Această afirmație este adevărată inițial, rămâne adevărată după fiecare pas și ne permite să demonstrăm corectitudinea la terminare.

> Un invariant nu descrie ce *sperăm* să se întâmple, ci o proprietate care rămâne adevărată pe tot parcursul algoritmului.

## 2. Căutarea binară

Dacă lista este sortată, comparăm ținta cu elementul din mijloc și eliminăm jumătate din interval după fiecare pas. După $k$ pași mai rămân aproximativ $n / 2^k$ elemente, deci numărul de pași este $O(\log n)$.

```python
def cautare_binara(valori, tinta):
    stanga, dreapta = 0, len(valori) - 1
    comparatii = 0

    while stanga <= dreapta:
        mijloc = (stanga + dreapta) // 2
        comparatii += 1
        if valori[mijloc] == tinta:
            return mijloc, comparatii
        if valori[mijloc] < tinta:
            stanga = mijloc + 1
        else:
            dreapta = mijloc - 1

    return -1, comparatii

date_sortate = list(range(0, 64, 2))
pozitie, pasi = cautare_binara(date_sortate, 22)
print("Poziție:", pozitie)
print("Comparații:", pasi)
```

### Urmărirea intervalului

Invariantul este: dacă ținta există, ea se află întotdeauna între indicii `stanga` și `dreapta`. Actualizarea limitelor păstrează invariantul deoarece lista este sortată. Fără această precondiție, eliminarea unei jumătăți nu este justificată.

```python
import math

dimensiuni = [16, 128, 1024, 65_536]
comparatie = [
    (n, n, math.ceil(math.log2(n)))
    for n in dimensiuni
]
comparatie
```

## 3. Alegerea strategiei

Căutarea binară este rapidă, dar cere date sortate. Sortarea costă de regulă $O(n \log n)$, deci nu merită să sortăm pentru o singură căutare. Pentru multe interogări pe aceleași date, costul inițial poate fi amortizat.

| Situație | Alegere rezonabilă |
|---|---|
| listă mică, o singură căutare | liniară |
| listă sortată, multe căutări | binară |
| multe inserări și căutări exacte | set / dicționar |
| date sortate, căutare de interval | binară (`bisect`) |
