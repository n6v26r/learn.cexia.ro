# Colecții și funcții

Programele utile lucrează cu mai multe valori și repetă aceleași transformări. Listele grupează date, iar funcțiile grupează instrucțiuni.

## 1. Liste

O listă este o colecție ordonată. Primul element are indicele `0`. Putem parcurge toate elementele cu o buclă `for`.

```python
scoruri = [7, 9, 10, 9]
media = sum(scoruri) / len(scoruri)
print("Media:", media)
print("Scor maxim:", max(scoruri))
```

## 2. Transformări

O list comprehension descrie concis o transformare. Mai jos normalizăm scorurile din intervalul 0–10 în intervalul 0–1, operație frecventă în machine learning.

```python
scoruri_normalizate = [scor / 10 for scor in scoruri]
scoruri_normalizate
```

## 3. Funcții

O funcție primește **parametri** și poate întoarce un rezultat prin `return`. Numele bun al unei funcții spune clar ce transformare realizează.

```python
def acuratete(predictii, etichete):
    corecte = sum(p == e for p, e in zip(predictii, etichete))
    return corecte / len(etichete)

print(acuratete([1, 0, 1, 1], [1, 0, 0, 1]))
print(acuratete([1, 1], [0, 0]))
```
