## 1. Valori și variabile

O **valoare** poate fi un număr, un text sau o valoare logică. O **variabilă** este un nume pe care îl asociem unei valori cu operatorul `=`.

| Exemplu | Tip | Utilizare |
|---|---|---|
| `17` | `int` | numere întregi |
| `3.14` | `float` | numere cu zecimale |
| `"CEXIA"` | `str` | text |
| `True` | `bool` | adevărat sau fals |

```python
nume = "Ada"
varsta = 14
print(f"Salut, {nume}!")
print(f"Peste 3 ani vei avea {varsta + 3} ani.")
```

## 2. Expresii și comparații

Python respectă ordinea operațiilor din matematică. Operatorii `>`, `<`, `==` și `!=` produc valori logice. Observă că `=` atribuie o valoare, iar `==` compară două valori.

```python
scor = 5 + 3 * 6
print(scor)
print(scor >= 20)
print(scor == 24)
```

## 3. Decizii cu `if`

O instrucțiune condițională execută ramura al cărei test este adevărat. Indentarea — cele patru spații din fața instrucțiunilor — face parte din sintaxa Python.

```python
temperatura = 23

if temperatura < 10:
    mesaj = "Este frig"
elif temperatura < 28:
    mesaj = "Temperatură plăcută"
else:
    mesaj = "Este cald"

print(mesaj)
```
