# Vectori, matrici și transformări

Un exemplu de date poate fi scris ca vector $x\in\mathbb{R}^d$. Un set de $n$ exemple devine matricea $X\in\mathbb{R}^{n\times d}$. Vom urmări formele obiectelor și sensul geometric al fiecărei operații.

## 1. Vectori, norme și similaritate

Produsul scalar $x^Ty=\sum_i x_i y_i$ măsoară alinierea. Norma euclidiană este $\lVert x\rVert_2=\sqrt{x^Tx}$, iar similaritatea cosinus este

$$\cos(x,y)=\frac{x^Ty}{\lVert x\rVert_2\lVert y\rVert_2}. $$

Aceasta compară direcțiile și elimină influența scării.

```python
import numpy as np

x = np.array([1., 2., 2.])
y = np.array([2., 1., 0.])
dot = x @ y
cosine = dot / (np.linalg.norm(x) * np.linalg.norm(y))
print('produs scalar:', dot)
print('similaritate cosinus:', round(cosine, 3))
```

## 2. Matricea ca transformare liniară

O matrice $W\in\mathbb{R}^{m\times d}$ transformă un vector din $\mathbb{R}^d$ în $\mathbb{R}^m$. Coloana $j$ arată contribuția coordonatei $x_j$, iar rândul $i$ definește o caracteristică de ieșire.

Pentru un batch stocat pe rânduri folosim $Y=XW^T+b$. Vectorul $b$ este adăugat fiecărui rând prin broadcasting.

```python
X = np.array([[1., 2., 0.], [0., 1., 1.], [2., 0., 1.]])  # (3, 3)
W = np.array([[1., .5, -.4], [.2, -.5, .8]])       # (2, 3)
b = np.array([.6, 1.0])                            # (2,)
Y = X @ W.T + b
Y
```

## 3. Rang, dependență și informație

Rangul este dimensiunea spațiului generat de coloane. Dacă $\operatorname{rank}(X)<d$, unele trăsături sunt combinații ale altora, iar parametrii unui model liniar pot să nu fie identificabili în mod unic.

Descompunerea SVD, $X=U\Sigma V^T$, ordonează direcțiile după cantitatea de variație capturată. Valorile singulare mici indică direcții aproape redundante.

```python
A = np.array([[1., 0., 1.], [0., 1., 1.], [1., 1., 2.]])
_, singular_values, _ = np.linalg.svd(A)
print('valori singulare:', np.round(singular_values, 3))
print('rang:', np.linalg.matrix_rank(A))
```

## 4. Forme pătratice și curbură

Expresia $x^TAx$ este o formă pătratică. Pentru o matrice simetrică pozitiv definită, toate valorile proprii sunt pozitive și suprafețele de nivel sunt elipse. Raportul $\kappa=\lambda_{max}/\lambda_{min}$ este numărul de condiționare: un $\kappa$ mare produce văi înguste și face optimizarea dificilă.
