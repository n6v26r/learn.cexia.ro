# Derivate, gradienți și regula lanțului

Antrenarea unui model înseamnă minimizarea unei funcții $L(\theta)$. Derivata descrie sensibilitatea locală, iar gradientul adună derivatele parțiale într-o direcție de creștere maximă.

## 1. De la variație la derivată

Definiția este

$$f'(x)=\lim_{h\to0}\frac{f(x+h)-f(x)}{h}. $$

În calcule numerice, diferența centrală are eroare de ordin $O(h^2)$: $f'(x)\approx[f(x+h)-f(x-h)]/(2h)$. Un $h$ prea mare introduce eroare de trunchiere, iar unul prea mic amplifică eroarea de rotunjire.

```python
def f(x): return x**3 - 2*x + 1
x, h = 1.0, 1e-5
numeric = (f(x + h) - f(x - h)) / (2*h)
print('analitic:', 3*x**2 - 2)
print('numeric:', numeric)
```

## 2. Gradientul unei funcții multivariate

Pentru $f(x,y)=x^2+3xy+2y^2$,

$$\nabla f(x,y)=\begin{bmatrix}2x+3y \\ 3x+4y\end{bmatrix}. $$

Derivata direcțională pe vectorul unitar $u$ este $D_u f=\nabla f^Tu$. Ea este maximă atunci când $u$ are direcția gradientului.

```python
import numpy as np
point = np.array([1., -1.])
grad = np.array([2*point[0] + 3*point[1], 3*point[0] + 4*point[1]])
u = grad / np.linalg.norm(grad)
grad, grad @ u
```

## 3. Regula lanțului ca graf computațional

Fie $z=wx+b$, $a=\tanh(z)$ și $L=(a-y)^2$. Derivăm din exterior spre interior:

$$\frac{\partial L}{\partial w}=2(a-y)(1-a^2)x. $$

Backpropagation aplică această regulă local în sens invers și reutilizează rezultatele intermediare. Costul tuturor derivatelor este de același ordin cu evaluarea funcției.

```python
x, y, w, b = 1.7, 0.4, -0.2, 0.1
def loss(w):
    a = np.tanh(w*x + b)
    return (a-y)**2
a = np.tanh(w*x + b)
analytic = 2*(a-y)*(1-a*a)*x
h = 1e-5
numeric = (loss(w+h)-loss(w-h))/(2*h)
relative = abs(analytic-numeric)/max(1e-12, abs(analytic)+abs(numeric))
print(f'gradient analitic: {analytic:.6f}')
print(f'gradient numeric:  {numeric:.6f}')
print(f'eroare relativă:   {relative:.2e}')
```

## 4. Jacobiană și Hessiană

Pentru $f:\mathbb{R}^n\to\mathbb{R}^m$, Jacobiana $J_{ij}=\partial f_i/\partial x_j$ descrie aproximația liniară locală. Hessiana $H_{ij}=\partial^2L/(\partial x_i\partial x_j)$ descrie curbura unei funcții scalare.

În deep learning nu construim de obicei aceste matrici dense. Produsele Jacobian-vector și Hessian-vector pot fi calculate implicit, evitând costul pătratic de memorie.
